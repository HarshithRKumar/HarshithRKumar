```python
class person:
  def __init__(self,name,age):
    self.name=name
    self.age=age

  def display(self):
    print(self.name,self.age)

p1=person("Harshi",22)
p1.display()

```

```python
class car: #class
  def __init__(self,brand,model):
    self.brand= brand
    self.model= model
  def display_info(self):
    print(f"car brand: {self.brand}, car model: {self.model}")

my_car=car("BMW","hytdj") #object
my_car.display_info()
```

```python
class human:
  def __init__(self,name,age):
    self.name=name
    self.age=age
  def be(self):
    print(f"My name is:{self.name}, and my age is:{self.age}")

person1=human('harshith',22)
person1.be()
```

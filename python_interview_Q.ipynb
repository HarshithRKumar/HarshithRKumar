```python
s="one two three four five six"
l=s.split()
l1=[]
i=0
while i<len(l):
  if i%2==0:
    l1.append(l[i])
  else:
    l1.append(l[i][::-1])
  i=i+1
print(' '.join(l1))
```

```python
s="B4A1D3"
alp=[]
dig=[]
for ch in s:
  if ch.isalpha():
    alp.append(ch)
  else:
    dig.append(ch)
output=(''.join(sorted(alp)+sorted(dig)))
print(output)
```

```python
s="a4b3c2"
output=''
for ch in s:
  if ch.isalpha():
    x=ch
  else:
    d=int(ch)
    output=output+x*d
print(output)
```

```python
s="aaabbccdee"
prev=s[0]
output=''
c=1
i=1
while i<len(s):
  if s[i]==prev:
    c=c+1
  else:
    output=output+prev+str(c)
    prev=s[i]
    c=1
  if i==len(s)-1:
    output=output+prev+str(c)
  i=i+1
print(output)
```

```python
s="a4k3b2"
output=''
for ch in s:
  if ch.isalpha():
    output=output+ch
    x=ch
  else:
    d=int(ch)
    new=chr(ord(x)+d)
    output=output+new
print(output)
```
